```python
import sys
print(sys.executable)
import requests
from bs4 import BeautifulSoup
```

```python
url = "https://pt.wikipedia.org/"
headers = {"User-Agent": "estudo-beautifulsoup/1.0 (projeto de faculdade)"}

resposta = requests.get(url, headers=headers)
print(resposta.status_code)
```

```python
soup = BeautifulSoup(resposta.text, "html.parser")
print(soup.title)
```

```python
h1 = soup.find("h1")
print(h1.get_text(strip=True))
if h1:
        print("H1:", h1.get_text(strip=True))
else:
        print("Nenhum H1 encontrado")
```

```python
primeiro_link = soup.find("a")
print(primeiro_link)
print(primeiro_link["href"])
```

```python
todos = soup.find_all("a")
print(len(todos))

for link in todos[:5]:
    print(link.get_text(strip=True), "->", link.get("href"))
```
